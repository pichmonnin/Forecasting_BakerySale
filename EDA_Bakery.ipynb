## Discussion
From this small EDA, I have seen that the data itself has no missing values or NaN values in any columns, but there is an unnecessary columns that I have to drop and change the name of column from 'article' -> 'Product_Name'

### Analyze Tier Product Purchase 
#### High Tier

```python
counts = df['Product_Name'].value_counts()
# High Tier
High_Tier= counts[counts >= 10000].index
df_high= df[df['Product_Name'].isin(High_Tier)]
df_high['Product_Name'].value_counts()
```

#### Mid Tier

```python
Mid_Tier = counts[(counts< 10000) & (counts >= 1000)].index
df_mid = df[df['Product_Name'].isin(Mid_Tier)]
df_mid['Product_Name'].value_counts()
```

#### Low Tier

```python
Low_Tier = counts[(counts < 1000) & (counts>= 100)].index
df_low = df[df['Product_Name'].isin(Low_Tier)]
df_low['Product_Name'].value_counts()
```

#### Rare Tier

```python
Rare_Tier = counts[counts < 100].index
df_rare = df[df['Product_Name'].isin(Rare_Tier)]
df_rare['Product_Name'].value_counts()
```

### Column Chart for the High Tier

```python
fig ,ax = plt.subplots(figsize = (10 ,5))
count_high  = df_high['Product_Name'].value_counts()
colors = sns.color_palette("Set2", n_colors = len(count_high.index))
bars = ax.bar(count_high.index , count_high.values , color = colors , width = 0.4)
for bar in bars:
  height = bar.get_height()
  formatted_text  = f"{height/1e3:.2f} K"
  ax.text(bar.get_x()  + bar.get_width()/2 , height*1.01 , formatted_text , ha='center' , va='bottom', fontsize= 12)
ax.spines[['right', 'top']].set_visible(False)
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda x, _ : f"{x/1e3} K"))
ax.spines['bottom'].set_color('#898989')
ax.set_axisbelow(True)
ax.yaxis.grid(True , color = '#898989')
ax.tick_params(axis= 'x', rotation =-45)
ax.tick_params(bottom =False , left = False)
ax.set_xlabel("High Tier Product", labelpad = 10 , fontsize =12)
ax.set_ylabel("Total Amount of Purchases", labelpad =15, fontsize =12)
ax.set_title("High Tier Purchase" , pad = 25 , weight ='bold' , fontsize =16)
plt.show()
```

# Discussion
From the EDA, we can see that **Traditional Baguette** generates the most sales, making it the bakery’s best-selling product. We can also use this dataset to forecast total sales of the top-tier products for the next three months.

```python
df[['year' , 'month','day']] =df['date'].str.split("-" , expand = True)
month = ['Jan', 'Feb', 'Mar' , 'Apr', 'May' , 'Jun', 'Jul', 'Aug', 'Sep','Oct','Nov', 'Dec']
df['month'] =df['month'].replace({'01':month[0], '02':month[1],'03':month[2],'04':month[3],'05':month[4],'06':month[5],'07':month[6],'08':month[7],'09':month[8],'10':month[9],'11':month[10],'12':month[11]})

## Total for each month in 2021
df_sale = df.pivot_table(index = 'Product_Name', columns = ['year','month'],values = 'sale', aggfunc='sum')
df_sale = df_sale.reindex(columns = month , level = 1)
df_sale
```
